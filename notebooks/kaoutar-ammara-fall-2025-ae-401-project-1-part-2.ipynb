## <a class="anchor" id="2._Introduction"></a>  2. Introduction 

[Back to Table of Contents](#Table_of_Contents) 

In this Jupyter Notebook, we employ the available literature on the ascent engine of the Apollo lunar module (LM) as a worked out example to practice the process of liquid fuel engine sizing. The ascent module was defined as "The most remarkable space vehicle ever" (Pucciarelli 2020) and its engine was even reconfigured to compete, though unsuccessfully, as the RS-18 engine running on methane and liquid oxygen as part of the then-canceled Constellation lunar exploration program (NASA 2008, Spaceref 2008, Wikipedia 2021, 2022). 

All needed technical data are available in the following publications linked below in the References: Bennett 1972, Grumman Aerospace Corporation 1971, Harmon 2009, Humphries and Taylor 1973, and Manned Spacecraft Center 1971. 

The main texts on rocket engine sizing to be considered are Humble 1995 (Ch. 3) and Turner 2009 (Ch. 2). In the same textbooks, chapters on liquid fuel engine design are very useful to consider the problem at hand. The text by Sutton offers an unparalleled, comprehensive treatment of all connected issues. 

Part of the initial performance study requires use of the Computer Program for Calculation of Complex Chemical Equilibrium Compositions and Applications program (NASA CEA) available both online and as a locally running program. Standard references are those by Gordon and McBride (1994, 1996) but two referenced YouTube videos also deliver useful introductory material. However, a very useful illustration of the process is also in Appendix B of Humble (1995).  

Part of this activity is not just to determine the correct answers but also to formulate the correct questions to be answered. In order to do so, the process of engine sizing must become clear.

The nozzle expansion ratio for the LM ascent engine is taken from historical data as:

$$
\frac{A_e}{A^*} = 46
$$

The exit Mach number is obtained from the isentropic area–Mach relation:

$$
\frac{A_e}{A^*}=\frac{1}{M_e}\left[\frac{2}{\gamma + 1}\left(1 + \frac{\gamma - 1}{2} M_e^2\right)\right]^{\frac{\gamma + 1}{2(\gamma - 1)}}
$$

Using a representative value of the ratio of specific heats for Aerozine-50/N$_2$O$_4$ combustion products,

$$
\gamma = 1.22
$$

```
gamma = 1.22;
AeAt = 46;
```

```
AreaMach[M_] =(1/M)*((2/(gamma + 1))* (1 + (gamma - 1)/2*M^2))^((gamma + 1)/(2*(gamma - 1)));
MeSolution = FindRoot[AreaMach[M] == AeAt, {M, 4}]
Me = M /. MeSolution
```

The exit-to-chamber pressure ratio is then computed from the isentropic pressure relation:

$$\frac{P_e}{P_c}=\left(1 + \frac{\gamma - 1}{2} M_e^2\right)^{-\frac{\gamma}{\gamma - 1}}$$

```
PressureRatio[M_] = (1 + (gamma - 1)/2*M^2)^(-gamma/(gamma - 1));
PePc = PressureRatio[Me]
PcPe = 1/PePc
PcPe // ScientificForm
```

Finally, the chamber-to-exit pressure ratio used in the analysis is:

$$
\boxed{\frac{P_c}{P_e} \approx 6.2 \times 10^{2}}
$$

```
Lstar= 0.78; 
Lc = Exp[0.029 *(Log[dstar 100])^2 + 0.47* Log[dstar 100] + 1.94] 0.01
```

```
Lchamber = 0.001 94.52;    (* equivalent conical length *)
Le=1.30223;

Rn=20.91 0.001; (* Throat radius of curvature [m] *)

Tn = 37.64 Degree;  (* Nozzle initial angle *)
Te = 10.46 Degree;  (* Nozzle exit angle *)


xChamberEnd = Lchamber;
xThroat = Lc;
xExit = Lc + Le;
```

```
(* Geometric Parameters *)

Dc  = 218.99*^-3;     (* Chamber diameter [m] *)
Dth = 109.50*^-3;     (* Throat diameter [m] *)
Dex = 980.35*^-3;    (* Exit diameter [m] *)

Rc  = Dc/2;
Rth = Dth/2;
Rex = Dex/2;

Lcyl = 94.52*^-3;    (* Cylindrical chamber length [m] *)
Lc   = 255.10*^-3;   (* Total chamber length [m] *)
Le   = 1302.23*^-3;  (* Nozzle length [m] *)

Rn = 20.91*^-3;      (* Throat radius of curvature [m] *)

(* Angles *)
Tn = 30 Degree;   (* Nozzle initial angle *)
Te = 12 Degree;   (* Nozzle exit angle *)


(* Axial Locations *)

xChamberEnd = Lcyl;
xThroat     = Lc;
xExit       = Lc + Le;

(* Radius Functions *)

(* Chamber cylinder *)
rChamber[x_] := Rc

(* Converging section *)
rConverging[x_] := Rc - (Rc - Rth)*((x - xChamberEnd)/(xThroat - xChamberEnd))^2*(3 - 2*(x - xChamberEnd)/(xThroat - xChamberEnd))

(* Parabolic bell nozzle *)
rNozzle[x_] :=Module[{xi = x - xThroat, L = Le},Rth + Tan[Tn]*xi + ((Rex - Rth - L*Tan[Tn])/L^2)*xi^2]

(* Combined Profile *)

radius[x_] := Piecewise[{{rChamber[x], 0 <= x <= xChamberEnd},{rConverging[x], xChamberEnd < x <= xThroat},{rNozzle[x], xThroat < x <= xExit}}]

(* 3D Engine Geometry *)

EngineNozzle =RevolutionPlot3D[{x, radius[x]},{x, 0, xExit},
  RevolutionAxis -> {1, 0, 0},
  PlotPoints -> 120,
  Mesh -> None,
  PlotTheme -> "ThickSurface",
  AxesLabel -> {"x (m)", "y (m)", "z (m)"},
  PlotLabel -> "Rocket Thrust Chamber and Parabolic Bell Nozzle",
  ViewPoint -> {1.4, -2.2, 1.1},
  AspectRatio -> 1]

Nozzle2D = Show[{ Plot[ radius[x], {x, 0, xExit}, PlotStyle -> Thin],
   Plot[0, {x, 0, xExit}, PlotStyle -> Dashed]},
  AspectRatio -> Automatic,
  AxesLabel -> {"x (m)", "r (m)"},
  PlotLabel -> "2D Rocket Nozzle Profile (Axisymmetric)",
  PlotRange -> All]

```
